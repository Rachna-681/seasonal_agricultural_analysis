**Cleaning the data**

```python
for col in ['Rainfall_mm', 'Soil_Moisture_pct', 'Yield_Tonnes_Ha']:
    df[col] = df.groupby('Season')[col].transform(lambda x: x.fillna(x.mean()))
```

```python

df.isnull().sum().sum()   # should now be 0
```

**Chart - Yield by season**

```python
season_order = ['Kharif', 'Rabi', 'Zaid']
colors = ['#2E7D32', '#F9A825', '#C62828']

vals = df.groupby('Season')['Yield_Tonnes_Ha'].mean().reindex(season_order)

plt.figure(figsize=(7,5))
bars = plt.bar(vals.index, vals.values, color=colors)
plt.title('Average Crop Yield by Season')
plt.ylabel('Yield (Tonnes/Hectare)')
for b in bars:
    plt.annotate(f'{b.get_height():.2f}', (b.get_x()+b.get_width()/2, b.get_height()),
                 ha='center', va='bottom')
plt.show()
```

Chart — Rainfall & Temperature by Season

```python
rain = df.groupby('Season')['Rainfall_mm'].mean().reindex(season_order)
temp = df.groupby('Season')['Avg_Temperature_C'].mean().reindex(season_order)

fig, ax1 = plt.subplots(figsize=(7,5))
ax1.bar(rain.index, rain.values, color='#1565C0', alpha=0.7, label='Rainfall (mm)')
ax1.set_ylabel('Rainfall (mm)', color='#1565C0')

ax2 = ax1.twinx()
ax2.plot(temp.index, temp.values, color='#D84315', marker='o', linewidth=3, label='Avg Temp (°C)')
ax2.set_ylabel('Avg Temperature (°C)', color='#D84315')
plt.title('Rainfall & Temperature Across Seasons')
plt.show()
```

C**hart — Resource usage (fertilizer & water)**

```python
fert = df.groupby('Season')['Fertilizer_kg_ha'].mean().reindex(season_order)
water = df.groupby('Season')['Water_Used_m3'].mean().reindex(season_order)

fig, axes = plt.subplots(1, 2, figsize=(11,5))
axes[0].bar(fert.index, fert.values, color=colors)
axes[0].set_title('Avg Fertilizer Use by Season'); axes[0].set_ylabel('kg/hectare')
axes[1].bar(water.index, water.values, color=colors)
axes[1].set_title('Avg Water Used by Season'); axes[1].set_ylabel('cubic meters')
plt.show()
```

**Chart — Profit & % loss-making farms by season**

```python
profit = df.groupby('Season')['Profit_INR'].mean().reindex(season_order)
loss_pct = df.groupby('Season').apply(lambda x: (x['Profit_INR'] < 0).mean() * 100).reindex(season_order)

fig, axes = plt.subplots(1, 2, figsize=(11,5))
bar_colors = ['#2E7D32' if v >= 0 else '#C62828' for v in profit.values]
axes[0].bar(profit.index, profit.values, color=bar_colors)
axes[0].axhline(0, color='black', linewidth=0.8)
axes[0].set_title('Average Profit by Season'); axes[0].set_ylabel('Profit (INR)')

axes[1].bar(loss_pct.index, loss_pct.values, color=['#F9A825','#EF6C00','#C62828'])
axes[1].set_title('% of Loss-Making Farms by Season')
for i, v in enumerate(loss_pct.values):
    axes[1].annotate(f'{v:.1f}%', (i, v), ha='center', va='bottom')
plt.show()
```

**Chart — Profit by Crop across Seasons**

```python
pivot = df.groupby(['Crop','Season'])['Profit_INR'].mean().unstack().reindex(columns=season_order)

pivot.plot(kind='bar', figsize=(10,6), color=colors)
plt.axhline(0, color='black', linewidth=0.8)
plt.title('Average Profit by Crop Across Seasons')
plt.ylabel('Profit (INR)')
plt.xticks(rotation=30, ha='right')
plt.show()
```

**: Correlation heatmap**

```python
cols = ['Rainfall_mm','Avg_Temperature_C','Humidity_pct','Soil_Moisture_pct',
        'Fertilizer_kg_ha','Water_Used_m3','Yield_Tonnes_Ha','Profit_INR','Disease_Pest_Risk_pct']
corr = df[cols].corr()

fig, ax = plt.subplots(figsize=(8,7))
im = ax.imshow(corr, cmap='RdYlGn', vmin=-1, vmax=1)
ax.set_xticks(range(len(cols))); ax.set_xticklabels(cols, rotation=45, ha='right')
ax.set_yticks(range(len(cols))); ax.set_yticklabels(cols)
for i in range(len(cols)):
    for j in range(len(cols)):
        ax.text(j, i, f'{corr.iloc[i,j]:.2f}', ha='center', va='center', fontsize=8)
plt.colorbar(im)
plt.title('Correlation Heatmap of Key Variables')
plt.show()
```

**Chart — Water efficiency by irrigation method**

```python
we = df.groupby('Irrigation_Method')['Water_Efficiency_t_per_1000m3'].mean().sort_values(ascending=False)

plt.figure(figsize=(7,5))
plt.bar(we.index, we.values, color=['#00838F','#6A1B9A','#2E7D32','#EF6C00'])
plt.title('Water Use Efficiency by Irrigation Method')
plt.ylabel('Tonnes per 1000 m³ water')
plt.show()
```

**Average profit by state**

```python
state_profit = df.groupby('State')['Profit_INR'].mean().sort_values(ascending=False)

plt.figure(figsize=(9,5))
plt.barh(state_profit.index[::-1], state_profit.values[::-1], color='#2E7D32')
plt.title('Average Profit by State')
plt.xlabel('Profit (INR)')
plt.show()
```

# Seasonal Agriculture Performance Dataset

This dataset contains detailed information about agricultural performance across various farms, crops, and seasons. It aims to provide insights into factors influencing crop yield, production, and profitability, enabling analysis of seasonal trends and regional variations in agricultural practices.

## Dataset Overview

*   **File Name**: `seasonal_agriculture_performance_dataset.csv`
*   **Number of Entries**: 4000
*   **Number of Columns**: 28

## Key Columns and Descriptions

The dataset includes a comprehensive set of variables, which can be broadly categorized as follows:

### Farm and Crop Identifiers
*   `Farm_ID`: Unique identifier for each farm.
*   `State`: The Indian state where the farm is located.
*   `District`: The district within the state.
*   `Crop`: The type of crop cultivated (e.g., Wheat, Maize, Chilli, Pulses, Rice, Cotton, Groundnut, Sugarcane).
*   `Season`: The agricultural season (Kharif, Rabi, Zaid).

### Farm Characteristics
*   `Farm_Area_Hectares`: Size of the farm in hectares.

### Environmental Factors
*   `Rainfall_mm`: Total rainfall in millimeters.
*   `Avg_Temperature_C`: Average temperature in Celsius.
*   `Humidity_pct`: Average humidity percentage.
*   `Sunlight_Hours_Day`: Average daily sunlight hours.
*   `Soil_pH`: Soil pH level.
*   `Soil_Moisture_pct`: Soil moisture percentage.

### Inputs and Practices
*   `Nitrogen_kg_ha`: Nitrogen applied in kg per hectare.
*   `Phosphorus_kg_ha`: Phosphorus applied in kg per hectare.
*   `Potassium_kg_ha`: Potassium applied in kg per hectare.
*   `Irrigation_Method`: Method of irrigation used (e.g., Rainfed, Drip, Sprinkler, Flood).
*   `Fertilizer_kg_ha`: Total fertilizer applied in kg per hectare.
*   `Pesticide_Litre_ha`: Pesticide applied in liters per hectare.
*   `Seed_Quality_Score`: A score indicating the quality of seeds used.
*   `Water_Used_m3`: Total water used in cubic meters.

### Performance and Output
*   `Yield_Tonnes_Ha`: Crop yield in tonnes per hectare.
*   `Production_Tonnes`: Total production in tonnes.
*   `Water_Efficiency_t_per_1000m3`: Water efficiency measured as tonnes of yield per 1000 cubic meters of water.
*   `Disease_Pest_Risk_pct`: Percentage risk of disease and pests.

### Financial Metrics
*   `Market_Price_INR_Tonne`: Market price in INR per tonne.
*   `Total_Cost_INR`: Total cultivation cost in INR.
*   `Revenue_INR`: Total revenue generated in INR.
*   `Profit_INR`: Profit earned in INR.

## Data Quality and Preprocessing

*   Missing values in `Rainfall_mm`, `Soil_Moisture_pct`, and `Yield_Tonnes_Ha` were imputed using the mean for each respective `Season`.
*   No duplicate rows were found in the dataset.

## Potential Analyses and Insights

This dataset can be used to:

*   **Analyze Seasonal Performance**: Investigate how different seasons (Kharif, Rabi, Zaid) impact crop yield, profitability, and resource utilization.
*   **Crop-Specific Studies**: Understand the optimal conditions and profitability for various crops.
*   **Impact of Environmental Factors**: Study the correlation between rainfall, temperature, humidity, sunlight, and soil conditions with agricultural output.
*   **Efficiency Analysis**: Evaluate the efficiency of different irrigation methods and the impact of fertilizer/pesticide usage.
*   **Profitability Assessment**: Identify factors driving profitability and analyze regions/crops that are most (or least) profitable.
*   **Risk Assessment**: Explore the relationship between environmental factors, practices, and disease/pest risk.

## Example Findings from Exploratory Data Analysis

*   **Kharif Season** generally shows the highest average crop yield and profitability, despite higher rainfall and moderate temperatures.
*   **Zaid Season** tends to be the least profitable, with a higher percentage of loss-making farms, often associated with higher temperatures and moderate water usage.
*   **Chilli and Sugarcane** appear to be highly profitable crops, particularly during the Kharif season.
*   **Rainfed and Drip irrigation methods** show higher water use efficiency compared to Sprinkler and Flood methods.
*   There's a notable **positive correlation between Rainfall, Yield, and Profit**, and also with Disease/Pest Risk, suggesting that while rainfall is crucial, it can also bring increased challenges.

# Seasonal Agriculture Performance Analysis

This project performs an exploratory data analysis (EDA) on a synthetic dataset related to agricultural performance across various farms, crops, and seasons. The goal is to uncover insights into factors influencing crop yield, production, and profitability, and to understand seasonal trends and regional variations in agricultural practices.

## Dataset

**File Name**: `seasonal_agriculture_performance_dataset.csv`

This dataset contains detailed information about agricultural performance, including:

*   **Farm Identifiers**: `Farm_ID`, `State`, `District`, `Crop`, `Season`
*   **Farm Characteristics**: `Farm_Area_Hectares`
*   **Environmental Factors**: `Rainfall_mm`, `Avg_Temperature_C`, `Humidity_pct`, `Sunlight_Hours_Day`, `Soil_pH`, `Soil_Moisture_pct`
*   **Inputs & Practices**: `Nitrogen_kg_ha`, `Phosphorus_kg_ha`, `Potassium_kg_ha`, `Irrigation_Method`, `Fertilizer_kg_ha`, `Pesticide_Litre_ha`, `Seed_Quality_Score`, `Water_Used_m3`
*   **Performance & Output**: `Yield_Tonnes_Ha`, `Production_Tonnes`, `Water_Efficiency_t_per_1000m3`, `Disease_Pest_Risk_pct`
*   **Financial Metrics**: `Market_Price_INR_Tonne`, `Total_Cost_INR`, `Revenue_INR`, `Profit_INR`

## Data Preprocessing

*   Missing values in `Rainfall_mm`, `Soil_Moisture_pct`, and `Yield_Tonnes_Ha` were imputed using the mean for each respective `Season`.
*   No duplicate rows were found.

## Key Findings from Exploratory Data Analysis

The analysis revealed several important insights:

*   **Kharif Season** generally exhibits the highest average crop yield and profitability, despite experiencing higher rainfall and moderate temperatures.
*   **Zaid Season** was found to be the least profitable, with a significant percentage of farms incurring losses. This season is characterized by higher temperatures and moderate water usage.
*   **Chilli and Sugarcane** crops showed high profitability, especially during the Kharif season.
*   **Rainfed and Drip irrigation methods** demonstrated superior water use efficiency compared to Sprinkler and Flood methods.
*   A notable **positive correlation exists between Rainfall, Yield, and Profit**, as well as with Disease/Pest Risk. This suggests that while rainfall is crucial for agricultural success, it also introduces increased challenges related to disease and pest management.
